```python
""" Практичні завдання:

1. Вивести повідомлення про рівень втрат противника на основі введеного числа.
2. Порахувати середній рівень готовності трьох підрозділів.
3. Створити цикл, що виводить день і номер сектора з номером розрахунку. """
```

```python
vtraty = int(input("Введіть рівень втрат противника (0-100): "))
if vtraty < 30:
    print("Втрати противника низькі.")
elif vtraty < 70:
    print("Втрати противника середні.")
else:
    print("Втрати противника високі.")
```

```python
pidrozdil_1 = int(input("Введіть рівень готовності підрозділу 1 (0-100): "))
pidrozdil_2 = int(input("Введіть рівень готовності підрозділу 2 (0-100): "))
pidrozdil_3 = int(input("Введіть рівень готовності підроз   ілу 3 (0-100): "))
print("Середній рівень готовності підрозділів:", (pidrozdil_1 + pidrozdil_2 + pidrozdil_3) / 3)    
```

```python
i = int(input("Введіть день тижня щоб дізнатись сектор 3 роти (1-7): "))   
if i == 1:
    print("Сектор 3 роти в понеділок: 1")
elif i == 2:
    print("Сектор 3 роти в вівторок: 3")
elif i == 3:
    print("Сектор 3 роти в середу: 2")
elif i == 4:
    print("Сектор 3 роти в четвер: 4")
elif i == 5:
    print("Сектор 3 роти в п'ятницю: 1")
elif i == 6:
    print("Сектор 3 роти в суботу: 3")
elif i == 7:
    print("Сектор 3 роти в неділю: 5")
else:
    print("Невірний день тижня.")
```
