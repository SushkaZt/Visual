```python
# 4. Вивести підрозділи, втрати яких перевищують 50 осіб.
losses = {'rer': 100, 'reb': 20, 'hbrya': 150}
for division, losses in losses.items():
    if losses > 50:
        print(f"Підрозділ {division} має втраи більше 50 осіб: {losses}")
```
