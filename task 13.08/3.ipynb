```python
def average_of_list(my_list):
    if len(my_list) == 0:
        return 0
    else:
        return sum(my_list) / len(my_list)  
```

```python
list1 = [10, 20, 47]
average_of_list(list1)
```

```python
# 1. Створіть функцію, яка на основі втрат і чисельності ворога оцінює результативність дій.
def rezultat(vtrat, chisel):
    if chisel <= 0:
        return "Чисельність ворога має бути більшою за 0"
    return (vtrat / chisel) * 100
a = int(input('Введіть кількість знищених: '))
b = int(input('Введіть чисельність ворога: '))

print(f"Результативність: {rezultat(a, b)}%")
```

```python
# 2. Напишіть функцію для обрахунку середніх втрат за кілька днів.
losses_by_day = [5, 10, 15, 20, 25] 
def average_losses(losses_by_day):
    if len(losses_by_day) == 0:
        return 0
    return sum(losses_by_day) / len(losses_by_day)  


losses = [5, 10, 15, 20, 25] 

print(average_losses(losses))
```

```python
# 3. Імпортуйте модуль `random` і змоделюйте напрямок наступу.
import random
# Є 4 можливі напрямки: "Північ", "Південь", "Схід", "Захід"
directions = ["Північ", "Південь", "Схід", "Захід"]
input("Натисніть Enter, щоб змоделювати напрямок наступу...")
print(f"Напрямок наступу: {random.choice(directions)}")


```

```python
# 4. Створіть власний модуль із функціями для оцінки логістичної готовності.
def logistic_readiness(fuel, supplies, ammunition, medical):
    return (fuel + supplies + ammunition + medical) / 4

palivo = float(input("Введіть кількість наявного палива у відсотках від загальної потреби (1-100): "))
supplies = float(input("Введіть кількість наявних приpasів у відсотках від загальної потреби (1-100): "))
bk = float(input("Введіть кількість наявних боєприpasів у відсотках від загальної потреби (1-100): "))
medic = float(input("Введіть кількість наявних медичних засобів у відсотках від загальної потреби (1-100): "))


print(f"Логістична готовність: {logistic_readiness(palivo, supplies, bk, medic)}%")
```
